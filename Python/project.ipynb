```python
df = df.drop(columns=['index'])
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(' ', '_')
    .str.replace('-', '_')
)
```

```python
pd.crosstab(
    df['fulfilment'],
    df['fulfilled_by'],
    dropna=False
)
```

```python
df['promotion_ids'].value_counts(dropna=False)
```

```python
df['promotion_ids'] = df['promotion_ids'].fillna('no id')
```

```python
df['fulfiled_by'] = df['fulfilled_by'].fillna('not applicable')
```

```python
df = df.loc[:, ~df.columns.str.contains('^Unnamed')]
```

```python
df['amount'].value_counts(dropna=False)
```

```python
df['currency'] = df['currency'].fillna('INR')
```

```python
df[df['amount'].isnull()]['status'].value_counts()
```

```python
df = df.dropna(subset=['amount'])
```

```python
df['status'].value_counts()
```

```python
df['courier_status'].value_counts(dropna=False)
```

```python
df[df['courier_status'].isnull()]['status'].value_counts()
```

```python
df['courier_status']=df['courier_status'].fillna('Cancelled')
```

```python
df[df['ship_city'].isnull()]['status'].value_counts()
```

```python

shipping_cols = ['ship_city', 'ship_country', 'ship_postal_code', 'ship_state']
df[shipping_cols] = df[shipping_cols].fillna('Unknown')
```
